```python
import requests
import pandas as pd

response = requests.get(
    "https://api.worldbank.org/v2/country/NG/indicator/NY.GDP.MKTP.CD?format=json&per_page=100"
)

data = response.json()
records = data[1]

df_gdp = pd.DataFrame(records)[["date", "value"]]
df_gdp.columns = ["year", "gdp_usd"]

df_gdp.dropna(inplace=True)
df_gdp["year"] = df_gdp["year"].astype(int)
df_gdp["gdp_usd"] = df_gdp["gdp_usd"].astype(float)

print(df_gdp.head())
print("GDP records:", len(df_gdp))
```

```python
rows = df_gdp.to_dict(orient="records")

response = supabase.table("nigerian_gdp").insert(rows).execute()

print("GDP data uploaded successfully!")
print("Rows uploaded:", len(rows))
```
